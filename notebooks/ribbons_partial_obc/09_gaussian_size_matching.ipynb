# 09 — Gaussian size matching for the partial-OBC ribbon

The original comparison uses Gaussian $R_G=3\sigma=2.5$ nm and cap footprint
$R_C=2.5$ nm. These have different half-height widths. Keep height 1.5 nm,
centre (5.025,1.5) nm and the periodic-x/open-y ribbon fixed. Compute two extra
Gaussian cases using the same Bloch solver and full settings as notebook 07.

For the exact cap, $r_{1/2}^2=R_C^2/2+h^2/4$. The nominal width match is
$$\sigma_{\rm width}=\sqrt{\frac{R_C^2/2+h^2/4}{2\ln 2}},\qquad R_G=3\sigma_{\rm width}.$$
This gives $R_G\simeq4.893$ nm. The second case matches the volume under the
height profile **on the actual ribbon**, including the open-edge truncation
and Gaussian periodic images; this gives $R_G\simeq4.079$ nm. This geometric
volume is not an electronic charge or a mechanical equilibrium constraint.

Both calculations use 81 Bloch momenta, 1201 energies in [-0.30,0.30] eV,
5 meV Gaussian broadening, intrinsic SOC 100 meV and zero Rashba. The cell
stays 10.050 x 10.212 nm; Gaussian image overlap is retained, not truncated.
Only the new widths are solved; existing flat, original-Gaussian and cap data
are reused. Matching width or volume does not match the cap's sharp rim.
